```python
from datasets import load_dataset
import json
import re
import unicodedata
import pandas as pd

# Cargar Killkan
ds = load_dataset("ctaguchi/killkan")

# Cargar JSON de traducciones
with open("kichwa_translations.json", "r", encoding="utf-8") as f:
    traducciones = json.load(f)

print("Ejemplos Killkan:", len(ds["train"]))
print("Traducciones JSON:", len(traducciones))
```

```python
def normalizar_texto(texto):
    if texto is None:
        return ""
    
    texto = texto.lower().strip()
    texto = unicodedata.normalize("NFKC", texto)
    
    # quitar puntuación
    texto = re.sub(r"[^\w\s]", "", texto)
    
    # quitar espacios múltiples
    texto = re.sub(r"\s+", " ", texto)
    
    return texto
```

```python
traducciones_norm = {}

for kichwa, espanol in traducciones.items():
    clave_norm = normalizar_texto(kichwa)
    traducciones_norm[clave_norm] = espanol

print("Diccionario normalizado listo:", len(traducciones_norm))
```

# RapidFuzzy

```python
from rapidfuzz import process, fuzz

claves_json = list(traducciones_norm.keys())

matches_final = []
sin_match_final = []

for ejemplo in ds["train"]:
    texto_kichwa = ejemplo["sentence"]
    texto_norm = normalizar_texto(texto_kichwa)

    # Match exacto
    if texto_norm in traducciones_norm:
        matches_final.append({
            "audio_path": ejemplo["audio"]["path"],
            "kichwa_original": texto_kichwa,
            "kichwa_normalizado": texto_norm,
            "kichwa_json_match": texto_norm,
            "espanol": traducciones_norm[texto_norm],
            "tipo_match": "exacto",
            "score": 100
        })
    else:
        # Match fuzzy
        mejor = process.extractOne(
            texto_norm,
            claves_json,
            scorer=fuzz.ratio
        )

        if mejor is not None:
            mejor_texto, score, _ = mejor

            if score >= 85:  
                matches_final.append({
                    "audio_path": ejemplo["audio"]["path"],
                    "kichwa_original": texto_kichwa,
                    "kichwa_normalizado": texto_norm,
                    "kichwa_json_match": mejor_texto,
                    "espanol": traducciones_norm[mejor_texto],
                    "tipo_match": "fuzzy",
                    "score": score
                })
            else:
                sin_match_final.append({
                    "audio_path": ejemplo["audio"]["path"],
                    "kichwa_original": texto_kichwa,
                    "kichwa_normalizado": texto_norm
                })
        else:
            sin_match_final.append({
                "audio_path": ejemplo["audio"]["path"],
                "kichwa_original": texto_kichwa,
                "kichwa_normalizado": texto_norm
            })

print("Matches totales:", len(matches_final))
print("Sin match:", len(sin_match_final))
```

```python
for i in range(min(20, len(matches_final))):
    print("Audio:", matches_final[i]["audio_path"])
    print("Kichwa Killkan:", matches_final[i]["kichwa_original"])
    print("Kichwa JSON:", matches_final[i]["kichwa_json_match"])
    print("Español:", matches_final[i]["espanol"])
    print("Tipo:", matches_final[i]["tipo_match"])
    print("Score:", matches_final[i]["score"])
    print("-" * 70)
```

```python
df_matches = pd.DataFrame(matches_final)
df_no_matches = pd.DataFrame(sin_match_final)

df_matches.to_csv("killkan_audio_kichwa_es.csv", index=False, encoding="utf-8-sig")
df_no_matches.to_csv("killkan_sin_match.csv", index=False, encoding="utf-8-sig")

print("Archivos guardados correctamente.")
```
